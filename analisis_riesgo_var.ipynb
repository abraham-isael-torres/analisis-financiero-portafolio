```python
import yfinance as yf
import numpy as np
import pandas as pd

# Descargar datos de COIN
data = yf.download("COIN", start="2024-01-01", end="2026-07-26")
retornos = data["Close"].pct_change().dropna()

retornos.head()
```

```python
confianza = 0.95
z_score = 1.645  # para 95% de confianza

# Extraer la columna como Serie (por si viene como DataFrame)
retornos_coin = retornos["COIN"] if isinstance(retornos, pd.DataFrame) else retornos

# Método 1: VaR paramétrico
media = retornos_coin.mean()
sigma = retornos_coin.std()
var_parametrico = media - z_score * sigma

# Método 2: VaR histórico
var_historico = retornos_coin.quantile(1 - confianza)

# Método 3: VaR Monte Carlo
np.random.seed(42)
simulaciones = np.random.normal(media, sigma, 10000)
var_montecarlo = np.percentile(simulaciones, (1 - confianza) * 100)

print(f"VaR paramétrico (95%, diario): {var_parametrico:.2%}")
print(f"VaR histórico (95%, diario): {var_historico:.2%}")
print(f"VaR Monte Carlo (95%, diario): {var_montecarlo:.2%}")
```

```python
confianza_99 = 0.99
z_score_99 = 2.326

var_parametrico_99 = media - z_score_99 * sigma
var_historico_99 = retornos_coin.quantile(1 - confianza_99)

print(f"VaR paramétrico (99%, diario): {var_parametrico_99:.2%}")
print(f"VaR histórico (99%, diario): {var_historico_99:.2%}")
```
